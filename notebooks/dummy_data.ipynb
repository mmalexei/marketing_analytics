```python
import pandas as pd
import numpy as np
```

```python
df_fb=pd.read_csv(r"..\data\01_facebook_ads.csv")
df_gg=pd.read_csv(r"..\data\02_google_ads.csv")
df_tt=pd.read_csv(r"..\data\03_tiktok_ads.csv")
```

```python
df_fb.rename(columns={
    "spend":"cost",
    "ad_set_id":"ad_group_id",
    "ad_set_name":"ad_group_name"
    }, inplace=True)

df_gg.rename(columns={
    "avg_cpc":"cpc"
},inplace=True)

df_tt.rename(columns={
    "adgroup_id":"ad_group_id",
    "adgroup_name":"ad_group_name"
},inplace=True)
```

```python
#FACEBOOK
df_fb["ctr"]=(
    df_fb["clicks"]/df_fb["impressions"]
    )

df_fb["cpc"]=(
    df_fb["cost"]/df_fb["clicks"]
)

df_fb["conversion_rate"]=(
    df_fb["conversions"]/df_fb["clicks"]
    )

df_fb["view_rate"]=(
       df_fb["video_views"]/df_fb["impressions"]
    )

#GOOGLE
df_gg["conversion_rate"]=(
    df_gg["conversions"]/df_gg["clicks"]
    )

#TIKTOK
df_tt["ctr"]=(
    df_tt["clicks"]/df_tt["impressions"]
    )

df_tt["cpc"] = (
    df_tt["cost"] / df_tt["clicks"]
    )

df_tt["conversion_rate"]=(
    df_tt["conversions"]/df_tt["clicks"]
    )

df_tt["view_rate"]=(
    df_tt["video_views"]/df_tt["impressions"]
    )
```

```python

def dummy_data(df,days):
    generated_data=[]
    for i in range(days):
        temp=df.copy()
        temp["date"]=(
            pd.to_datetime(temp["date"])
            + pd.Timedelta(days=i+1)
        )

        temp["impressions"]=(
            temp["impressions"]*np.random.uniform(0.92,1.08,len(temp))
        ).astype(int)

        temp["ctr"]=(
            temp["ctr"]*
            np.random.uniform(0.97,1.03,len(temp))
        ).round(4)

        temp["clicks"]=(
             temp["impressions"]*temp["ctr"]
        ).astype(int)

        temp["cpc"]=(
            temp["cpc"]*
            np.random.uniform(0.97,1.03,len(temp))
        ).round(2)

        temp["cost"]=(
            temp["clicks"]*temp["cpc"]
        ).astype(int)

        temp["conversion_rate"]=(
            temp["conversion_rate"]*
            np.random.uniform(0.95,1.05,len(temp))
        ).round(4)
        
        temp["conversions"]=(
            temp["clicks"]*temp["conversion_rate"]
        ).astype(int)

        if "video_views" in temp.columns:
            temp["view_rate"]=(
                temp["view_rate"]*
                np.random.uniform(0.95,1.05,len(temp))
            ).round(4)

            temp["video_views"]=(
                temp["impressions"]*temp["view_rate"]
            ).astype(int)

        if "reach" in temp.columns:
            temp["reach"]=(
                temp["impressions"]/
                np.random.uniform(1.10,1.30,len(temp))
            ).astype(int)

        if "frequency" in temp.columns and "reach" in temp.columns:
            temp["frequency"]=(
                temp["impressions"]/temp["reach"]
            ).round(2)
        
        if "engagement_rate" in temp.columns:
            temp["engagement_rate"]=(
                temp["engagement_rate"]*
                np.random.uniform(0.95,1.05,len(temp))
            ).round(4)

        generated_data.append(temp)

    return pd.concat(generated_data,ignore_index=True)
```

```python
fb_dummy=dummy_data(df_fb,90)
gg_dummy=dummy_data(df_gg,90)
tt_dummy=dummy_data(df_tt,90)
```
